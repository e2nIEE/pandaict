# pandaICT Requirements, Design and Development Handbook

This pages gives an overview of the requirements, architectural design principles and development goals for pandaICT as developed in project ReCoDe. 


## Requirement analysis - Background

### ReCoDe Project Description

The following information is compiled from the ReCoDe "Gesamtvorhabensbeschreibung", version 20220928/final. ReCoDe generally aims to develop a Co-Simulation platform for examination of use cases in digitalized energy systems, including communication and market mechanisms. This is to be done by consequent further development of existing tools, i.a. pandapower, mosaik (OFFIS), SIMONA (CNI/TUDo) and OMNet++/ns-3, and enabling an "integrated simulation of information- and communication technologies". Comprehensive installation manuals, process descriptions and example configurations are to be given.

The project description states that UNIK and IEE have developed a **pandaICT** concept for considering the effects of communication infrastructures onto grid operation (apparently, this sounds like there is a final concept). pandaICT is to be further developed in ReCoDe to achieve a wider range of applicability for **different types of ICT analyses**. Furthermore, IEE aims at aquiring competencies for counseling services regarding **ICT and grid operation security and robustness**. Also, pandaICT is to be **coupled with software from project partners, i.e. SIMONA, mosaik and OMNet++/ns-3**, to create a holistic tool for e.g. researching agent-based grid operation with consideration of ICT networks and their effects. The tool is to be used for researching big, multiple grid-level and grid-operator scenarios (Innovation 1) which are to be made **open source towards the research community** (Innovation 2), but be available for commercial use through a **"Freemium" license model** to be defined by the project (AS 1.2) in collaboration with openKonsequenz. 

Netze BW, EWE Netz and e-Netz are associated partners to the project. With those, A concrete scenario and use case is to be defined and project results are to be used for researching operation of controllable loads acc. to §14a along the communication chain between grid control center and smart meter gateways (AS 1.2). The goal should be to determine the effects of communication latency fluctuations onto deration of controllable local systems in case of grid overload situations. Also, robustness and stability (e.g. unwanted oscillation effects) shall be researched. Used methods and software are to be sound towards validation and liability as the use case is security relevant. Related this is AP 2 where workshops for requirements and use case definitions are organized. IEE should derate pandaICT requirements from this. 

Possible interfaces between the software tools provided by ReCoDe partners are conceptualized in AP3. pandaICT is, together with OpSim, one of the IEE tools that should be interfaced with. 

The actual implementation work is done in AP 4. Here in AS 4.4, CNI considers an open source modelling of diverse ICT technologies, apparently with a focus on radio networks, heterogenous and resilient ICT networks. Furthermore, in AS 4.5 CNI considers shared communication infrastructures. Feature development of pandaICT happens in AS4.9 with 6 PM for IEE until end of Aug 2024 (current time plan). The goal is creating a tool for modeling of different ICT structures for **arbitrary pandapower networks**, as well as a **user documentation**. The AP5 (tool integration) description states that pandaICT allows for **reliability analysis**. In AP6, additional scenarios with SimBench grids and pandaICT models may be developed and in AP7, all software tools are integrated by Co-simulation. AP8 is about reference scenario calculations. For the latter APs, pandaICT generally needs to be provided and applied.

## Requirements

### Technical Requirements

- The product must be able to calculate ICT system reliability and total outage probability
- There is no need to support time series calculation at this point
- The product API should be designed such that simple interfacing to OpSim is possible. This doesn't seem too much of a challenge as long as pandapower's internal timeseries calculation is not used. Modelling or loading an ICT network model and conducting static statistical evaluations should be possible using a Python OpSim client as a wrapper both for OpSim-pandaICT and e.g. OpSim-Mosaik. 
- pandaICT model components must at some point interface to the electrical pandapower model. The data end devices, namely generators and loads, are natural coupling points in order to model which assets are communicating. But the exact design of this is to be developed considering the use case(s).
- The question how the pandaICT model should be integrated in the pandapower model is to be clarified. Should the ICT model be separate from the electrical model, or should it all be integrated into a single pandapower "net" object? 
- The product needs to model and support a variety of communication protocols on field and utility level, namely IEC 61850, IEEE 104, Tase.2 on the utility level and e.g. MODBUS or EEBus on the field level. Also the ones considered in AS 4.4.
- The product needs to model communication links / protocols that allow for quality of service definition and high-priority message transport as used in high-resilience networks, and defined e.g. in 5G or IEC 1850 (Goose). 
- Shared networks / network slicing (AS 4.5) should also be considered

### Development Process Requirements

- Development must adhere to the software quality management as defined by Benjamin
- Development should be use case driven to support the goal of modelling a SimBench based reference scenario that spans multiple grid levels. Core architecture definition should base on the reference use case.
- Gitlab is to be used


### Non-technical Requirements

- The product needs to support the licensing model defined in ReCoDe. This includes being open source towards research, but partly closed source towards commercial users. Thus, licensing conditions for each external library used must be checked versus this requirement. 
- It should be defined which functions are considered for the "premium" license.
- Licensing needs to be harmonized with ReCoDe AS1.2
- Compatibility of the product to adjacent developments in ReCoDe is to be checked continuously. 
- The reference use case should match one or multiple use cases developed in AP2
