# MTBF and MTTR for interconnected systems

Generally, pandaICT networks consist of elements with individal MTBF and MTTR that are interconnedted. First of all, the terms are defined as such: 

- MTTR is the mean time to repair an element after a failure occured. This includes failure detection, diagnosis, repair and bringing back online.
- MTTF (which is sometimes used) is the mean time an element runs after repair until it fails. 
- MTBF is the mean time between occurence of element failures. As seen from the diagram, MTBF = MTTF + MTTR.
- Availability (A) is the mean ratio between the element uptime and total time (which is uptime+downtime).
- Reliability (R) according to [5, 2.5 and 2.6] can be specified in various ways, e.g. using the MTBF

Notably, MTTF and MTBF are only relevant for elements that are repairable [5], which we generally assume to be the case for ICT network components. 
Also, it's sufficient to consider two paramaters of the set [MTTR, MTTF, MTBF] because the third can be calculated from any two. 
Availability on the other hand, using the definitions above, can be expressed in different ways: 

$$
    A = \frac{MTTF}{MTBF}=\frac{MTBF-MTTR}{MTBF}=\frac{MTTF}{MTTF+MTTR}
$$

We'll work with MTBF and MTTR from here on. Now, how can those be calculated for the whole system? Researching this shows that such calculation is different for serial and parallel systems, and different sources give different formula. For serial systems, the German Wikipedia page [1] gives a formula different from the English version [2] as well as Aurora Cons. [3] and Legrand [4]. For parallel systems, the calculation depends on how many of them need to fail for the whole system to fail. Most often looked at is the case that all parallel subsystems have to fail for that. For this, the German Wiki [1] gives a different formula than the English one [2] and Aurora [3], while Legrand [4] again gives another one. [5] on the other hand is the most concise source and calculates system MTBF and MTTR using reliability and availability. 

This gave me a considerable amount of confusion - especially finding [1]-[4] first, I couldn't explain the differences to myself. However, I now think that the German Wiki suffers from using MTBF when in fact they mean MTTF, while the English Wiki [2] seems to be consistent with Aurora [3] - hence I trust the English Wiki more. Still, let's have a closer look at [5] because it is most concise. 

According to [5] (and also others I've seen), the main characteristics for reliability is that it degrades over time, which is expressed by the function: 

$$
    R(t) = e ^ {-\lambda t}
$$

.. where $\lambda$ is a kind of measure for the degradation speed. [5] calls $\lambda$  "failure rate". In case of repairable systems, it would seem that the usual relation between R, \lambda and MTBF is: 

$$
    MTBF = \int_0^\infty e ^ {-\lambda t} dt = \frac{1}{\lambda}
$$

These definitions are consistently used by [2],[3] and [4]. Hence the formula for serial MTBF according to [5] matches [2] and [4]. For parallel systems, [5] does not give a well-useable formula but [2] and [4] do (their formula are consistent, too). For serial MTTR, the same is the case. For parallel MTTR, all three sources are again consistent.
