# Validation of pandaICT

As of now, the tests conducted with pandaICT focused on functionality (does the code run?), reproducability (will the results stay the same if the code is futher developed?) and plausibility (do the results make any sense`). The latter aspect was judged manually for a limited number of simple test networks.  

What's missing so far is a correctness check, i.e. validation. In some cases, this is not trivial: e.g. for the IP test network, latencies on connection level are calculated depending on the route of packages through the network and individual link latencies, which is tedious to calculate by hand, 

Hence, validation needs another tool which can be used to model the same networks and calculate the same results. The tool should support shortest-path routing, and cover an as wide as possible range of pandaict computations. The requirements are detailed below.
