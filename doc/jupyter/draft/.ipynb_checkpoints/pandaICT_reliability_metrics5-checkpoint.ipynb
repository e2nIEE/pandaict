# Application to pandaICT

How can we apply the learnings so far to a pandaICT network? 
Considering a single connection and a specific communication direction (down- or upstream), multiple paths might be on the route between communication peers. Link usage probability (LUP) tables express the available routes for a connection, and state which percentage of information items take them, which depends on the routing algorithm selected. In any case, one might draw a reliability block diagram that looks somewhat like this (where the end device might be replaced by a CONVERTER-type communication node):



Note that the nodes shown here might not be unique, i.e. the same node might appear in different routes. p1 ... pr denote the probabilities with which information items take one of the routes 1..r. The Nodes are not given in the LUP tables, but can be easily calculated by traversing the links along the route. 

Now, this looks like we generally have a mixed system consisting of serial subcomponents (the routes) that are connected in parallel. However, the graph is flawed because some nodes in there might re-appear in multiple routes. The real ICT network might be much more complex, and can not be depicted in this way. However, there is one exception to this: **if each information item uses exactly one route, i.e. information items are **not** copied within the network, parallel subsystem connections do not play out**. Instead, the total MTTR and MTBF are a combination of the serial MTTR and MTBF. This is the case for most routing algorithms (shortest-path, random, lowest-loading, round-robin). In fact, I can only fathom one routing algorithm where it is not the case at this time, which is flooding. 

Hence it seems to make sense to implement our calculations for this case, which we call "single-path / no-copy routing".
