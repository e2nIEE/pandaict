# How to use pandaICT's reliability module to calculate k-out-of-n (multi-source-target) scenarions
As discussed in `01_how_to_MTBF` the connection element defines under which circumstances we consider the network as functional. Up until now we have exclusively considered networks where we require an endpoint $A$ to be able to send messages to another endpoint $B$. But one could also be interested in more complex scenarions, where $A$ wants to talk to $B$ **and** $B$ wants to talk to $C$. Or even $A_1$ **or** $A_2$ have to talk to $B$. We could even think of scenarios where we have $n$ sources $A_1$, $A_2$, ..., $A_n$ and want $k$ of those to communicate with $B$. All those scenarios and more are natively supported by pandaICT and in this notebook we will learn how:

```python
import pandaict as pi
from pandaict import inspect
from pandaict.utils.enums import *
from pandaict.reliability.reliability import estimate_mtbf_mttr_for_conn
from demo_networks import simple_multi_net
```

```python
# ===========================================================
# simple_multi_net
#
# {appA}~~~~~~~~~~~~~~~~~~~{appB}~~~~~~~~~~~~~~~~~~~{appC}
#   |                        |                        |
# [Alice] --- (nodeAB) --- [Bob] --- (nodeBC) --- [Charlie]
#
# []=dev, ()=node, {}=app
# ===========================================================

net = simple_multi_net()
```
