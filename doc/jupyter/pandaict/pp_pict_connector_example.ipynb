*Jan Ringelstein*

This document details a reference example network for the pandapower-pandaICT connector (pp-pict-connector). For basic information about pandaICT modelling (nodes, links etc.), please refer to the jupyter notebook on datamodel before reading this.

The ICT network shown here is part of pandaICT's networks module. It can be created with the function create_testnet1(). More information about the SimBench network used, a possible use case and scenario can be found in the jupyter notebook on pandaICT basic scenario and functionalities.

## Reference Network Example 

The following graphic shows the reference network example as pandapower network with a physical pandaICT/pp_pict_connector network on top. 



The network includes a grid operator control center (left side, "CC") that is connected over different nodes, links and converters to a Wind power plant and a PV plant (right side). The generators are equipped with ICT devices that represent power controllers ("PC"). Nodes are placed at pp buses 1 and 3. Also, there are connections to two PQ meters (middle, at bus 5, "PQ") which are supposed to be situated in a HV/MV substation together with Node 2. There is a switch controller "SC" at bus 7. Note that there are three converter-type devices for protocol conversion: Node 3 is situated at bus 6 and has a LAN connection to the Wind power plant controller, Node 4 is situated at the PV plant main connection cabinet and is connected to its controller by a local network. Node 5 is situated at the bus 7 substation, connecting to the switch controller by a binary link.

The next graphic additionally shows the logical pandaICT/pp_pict_connector network. Connections are shown in green, protocols are given as tags. Note that the **Apps** are not explicitly shown, but one app is present at each connection endpoint. The Control Center may generally host multiple apps, but currently it is modeled with a single one. On each converter-type device an app is running which is specified to convert between the named protocols. 



The information items are defined according to the underlying use case, which foresees the control center receiving measurements for power from the power controllers and PQ meters, sending target values for active and reactive powers to the power controllers, as well as a target value to the switch controller. PQ meters and wind power plant use both active and reactive power information items, whereas the PV plant only uses active power. Furthermore, the switch position is communicated from the switch controller to the CC, and a switch position target vice-versa.

The setup could thus be used for the CC to perform active or reactive power curtailment as well as network topology change in case of network congestion as indicated by the PQ meters. There are timeseries for Wind and PV plants available for this use case.
