# pandaICT development tools

## Jupyter notebook

Jupyter notebooks for pandaICT documentation can be locally written using the `jupyter notebook` command. Images can be copied and pasted in there, and are automatically converted to be correctly displayed when the notebook is uploaded into the git. 

You can *not* use `raw` boxes in jupyter notebooks though, since they are *not* displayed at all in the git-created notebooks. Instead of `raw`, use `code`. While doing so, avoid that error messages appear in your local version. I've switched off the ability to execute code boxes in the git.created html version, so this should be fine. 

The url's for accessing the git-created Jupyter notebooks can be found in the pandaict channel of our ReCoDe team.  

## Sphinx and ReadTheDocs

Sphinx and ReadTheDocs are used by the pandapower team to create their beautiful online documentation under https://pandapower.readthedocs.io. It might be wise to use the same tools in order to seamlessly integrate the pandaICT documentation there, or setup a separate pandaICT online doc with the same look&feel.

**Sphinx** is a documentation generator that can translate a set of reStructuredText documents to a series of HTML files. Cross-references, indices, links etc. are produced automatically. Under Windows, it can be [installed](https://www.sphinx-doc.org/en/master/usage/installation.html#windows-other-method) with admin priviledges using pip. Install can be verified like this:


```python
pip install -U sphinx
sphinx-build --version
```

(in my case it was already installed). 

The [ReadTheDocs theme for Sphinx](https://sphinx-rtd-theme.readthedocs.io/en/stable/) is installed by:

```python
pip install sphinx_rtd_theme
```

In the Sphinx project’s `conf.py` file, `sphinx_rtd_theme` has to be added to the list of enabled extensions and set as the active theme (but for our doc, I already did that, so things should run right from the git):

```python
extensions = [
    ...
    'sphinx_rtd_theme',
]

html_theme = "sphinx_rtd_theme"
```

I've placed the sphinx root folder for pandaICT at the git under recode/pandaict/doc/sphinx. In there the doc can be built using:

```python
sphinx-build -M html source/ build/
```

The resulting html is under the build/ subdirectory.

Finally, [ReadTheDocs](https://about.readthedocs.com/) is apparently an online documentation building & hosting service. It offers a free community plan for open source projects. It would seem that it can directly work with Sphinx projects, and that we just need an account to upload our stuff or ask the pandapower team to add our doc to their site. ReadTheDocs can also be linked to a github, in which case the public online doc is re-generated and updated on every commit. 

## Tips & Tricks for Sphinx

Sphinx can be awkward at times. Here's what I found:

- In the python code sphinx comments, you can use a backslash \ at the end of a line to indicate that the following line should be appended to the previous one. BUT: there must not be any character behind the backslash - expecially no space - or sphinx will mess it up in html.
- In the toctree directives in .rst files, avoid using maxdepth:0 because it causes elements in the html left sidebar disappearing
