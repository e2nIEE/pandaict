# pandaICT development tools

## Sphinx and ReadTheDocs

Sphinx and ReadTheDocs are used by the pandapower team to create their beautiful online documentation under https://pandapower.readthedocs.io. It might be wise to use the same tools in order to seamlessly integrate the pandaICT documentation there, or setup a separate pandaICT online doc with the same look&feel.

**Sphinx** is a documentation generator that can translate a set of reStructuredText documents to a series of HTML files. Cross-references, indices, links etc. are produced automatically. The resulting html is under the build/ subdirectory.

Finally, [ReadTheDocs](https://about.readthedocs.com/) is apparently an online documentation building & hosting service. It offers a free community plan for open source projects. It would seem that it can directly work with Sphinx projects, and that we just need an account to upload our stuff or ask the pandapower team to add our doc to their site. ReadTheDocs can also be linked to a github, in which case the public online doc is re-generated and updated on every commit.
